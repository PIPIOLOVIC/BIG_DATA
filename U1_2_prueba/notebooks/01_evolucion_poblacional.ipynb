### Actividad 3 — Estadística descriptiva

#### Propósito
Aplicar medidas estadísticas básicas a una variable poblacional.

#### Instrucciones
Seleccionamos el año histórico **2020** y calculamos las medidas estadísticas descriptivas sobre la variable `Annual change in population` utilizando Pandas y NumPy.

```python
# Seleccionar un año del periodo histórico y filtrar los datos correspondientes
year_selected = 2020
df_year = df[df["Year"] == year_selected].copy()
col_target = "Annual change in population"

# Medidas estadísticas descriptivas
media = df_year[col_target].mean()
mediana = df_year[col_target].median()
varianza = df_year[col_target].var()
desviacion = df_year[col_target].std()
minimo = df_year[col_target].min()
maximo = df_year[col_target].max()

# Identificar entidades con valor mínimo y máximo
entidad_min = df_year.loc[df_year[col_target].idxmin(), "Entity"]
entidad_max = df_year.loc[df_year[col_target].idxmax(), "Entity"]

print(f"--- Estadísticas descriptivas para el año {year_selected} ---")
print(f"Media:                {media:,.2f}")
print(f"Mediana:              {mediana:,.2f}")
print(f"Varianza:             {varianza:,.2f}")
print(f"Desviación estándar:  {desviacion:,.2f}")
print(f"Mínimo:               {minimo:,.2f} (Entidad: {entidad_min})")
print(f"Máximo:               {maximo:,.2f} (Entidad: {entidad_max})")
```

#### Preguntas de la Actividad 3 (para el año 2020)
1. **¿Cuál es el valor de la media?**  
   La media es **2,124,908.85** personas.

2. **¿Cuál es el valor de la mediana?**  
   La mediana es **34,293.50** personas.

3. **¿Cuál es el valor de la desviación estándar?**  
   La desviación estándar es **9,522,989.82** personas.

4. **¿Qué entidad presenta el valor mínimo y cuál el máximo?**  
   - **Valor mínimo:** **Venezuela** con **-494,016** personas.
   - **Valor máximo:** **World** con **75,707,590** personas (a nivel país individual, **India** con 13,871,885 personas).
